```sql
-- Create catalog
CREATE CATALOG fraudwatch
COMMENT 'Catalog for fraudwatch project'
;
```

```sql
-- Create schemas for the catalog
CREATE SCHEMA IF NOT EXISTS fraudwatch.bronze;
CREATE SCHEMA IF NOT EXISTS fraudwatch.silver;
CREATE SCHEMA IF NOT EXISTS fraudwatch.gold;
```

```sql
-- Create schema to store etl events, metadata,...
CREATE SCHEMA IF NOT EXISTS fraudwatch.etl;
```

```sql
-- Create external location for watchlist
CREATE EXTERNAL LOCATION ext_watchlist
URL "abfss://data@fraudwatch01.dfs.core.windows.net/watchlist"
WITH (STORAGE CREDENTIAL fraudwatch_sc)
;
```

```sql
-- Create a volume to store watchlist streaming files
CREATE EXTERNAL VOLUME IF NOT EXISTS fraudwatch.source.watchlist
LOCATION 'abfss://data@fraudwatch01.dfs.core.windows.net/watchlist'
;
```
